# 🔹 What Is RMSProp?

RMSProp is an adaptive learning rate optimizer that fixes AdaGrad’s problem by using an exponentially weighted moving average of squared gradients.

👉 It prevents the learning rate from shrinking too fast.

# 🔹 Why RMSProp Was Introduced

AdaGrad problem:

Accumulates all past squared gradients

Learning rate → almost zero

RMSProp solution:

Uses recent gradients only

Forgets old gradients gradually

# 🔹 Core Idea (Simple Intuition)

Large gradients → smaller learning rate

Small gradients → larger learning rate

But only based on recent history.

# 🔹 Mathematical Formulation
EWMA of Squared Gradients
sₜ = β sₜ₋₁ + (1 − β) gₜ²

Weight Update
wₜ = wₜ₋₁ − η gₜ / √(sₜ + ε)


Where:

gₜ = current gradient

η = learning rate

β ≈ 0.9

ε = small constant

# 🔹 Key Characteristics
Feature	RMSProp
Learning rate	Adaptive
Memory	EWMA
Best for	RNNs, non-stationary problems
Convergence	Fast & stable
# 🔹 Advantages

✅ Fixes AdaGrad’s decay problem
✅ Faster convergence than SGD
✅ Works well for RNNs
✅ Less tuning required

# 🔹 Disadvantages

❌ No momentum term (basic RMSProp)
❌ May not generalize as well as SGD

# 🔹 RMSProp vs AdaGrad

| Feature             | AdaGrad     | RMSProp        |
| ------------------- | ----------- | -------------- |
| Gradient memory     | All history | Recent history |
| Learning rate decay | Too fast    | Controlled     |
| Deep learning       | ❌           | ✅              |


# 🔹 RMSProp vs Adam

| RMSProp                | Adam                          |
| ---------------------- | ----------------------------- |
| Uses squared gradients | Uses mean + squared gradients |
| No bias correction     | Bias correction               |
| Simpler                | More powerful                 |

# 🔹 RMSProp in Keras

from tensorflow.keras.optimizers import RMSprop

optimizer = RMSprop(
    learning_rate=0.001,
    rho=0.9
)

model.compile(
    optimizer=optimizer,
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

🧠 One-Line Exam Answer

RMSProp adapts the learning rate using an exponentially weighted moving average of squared gradients, preventing rapid decay of learning rates.

# 🔹 What Is Adam?

Adam is one of the most popular optimizers in deep learning.
It combines the best ideas of:

Momentum → fast convergence

RMSProp → adaptive learning rate

👉 Adam = Momentum + RMSProp

# 🔹 Why Adam Is So Popular

Works well out of the box

Little hyperparameter tuning

Handles noisy & sparse gradients

Fast and stable training

# 🔹 Core Idea (Intuition)

Adam keeps track of:
1️⃣ Mean of gradients (direction)
2️⃣ Variance of gradients (step size)

Both are computed using Exponentially Weighted Moving Averages (EWMA).

# 🔹 Mathematical Formulation
Step 1: Compute Gradients
gₜ = ∇L(wₜ)

Step 2: First Moment (Momentum)
mₜ = β₁ mₜ₋₁ + (1 − β₁) gₜ

Step 3: Second Moment (RMSProp)
vₜ = β₂ vₜ₋₁ + (1 − β₂) gₜ²

Step 4: Bias Correction
m̂ₜ = mₜ / (1 − β₁ᵗ)
v̂ₜ = vₜ / (1 − β₂ᵗ)

Step 5: Update Weights
wₜ₊₁ = wₜ − η · m̂ₜ / (√v̂ₜ + ε)

# 🔹 Default Hyperparameters (Important ❗)
Parameter	Value
Learning rate (η)	0.001
β₁	0.9
β₂	0.999
ε	1e−8
# 🔹 Advantages

✅ Fast convergence
✅ Adaptive learning rate per parameter
✅ Works well for deep networks
✅ Handles sparse gradients

🔹 Disadvantages

❌ May generalize worse than SGD
❌ Can overfit
❌ Sometimes unstable without tuning
