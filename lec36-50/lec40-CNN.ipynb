# What is a Convolutional Neural Network (CNN)?

A Convolutional Neural Network (CNN) is a type of deep learning model specially designed to process images and spatial data by automatically learning visual features like edges, shapes, textures, and objects.

👉 CNNs are the backbone of computer vision.

# 🔹 Why CNNs Are Needed

Traditional neural networks:

Treat images as flat vectors

Lose spatial information

Have too many parameters

CNNs solve this by:

Preserving spatial structure

Sharing weights

Reducing parameters

# 🔹 How CNN Works (Big Picture)
Image → Convolution → Activation → Pooling → Fully Connected → Output

# 🔹 Core Components of CNN
# 1️⃣ Convolution Layer (Heart of CNN ❤️)

Uses small filters (kernels)

Slides over the image

Detects features (edges, corners)

📌 Output: Feature maps

# 2️⃣ Activation Function (ReLU)

Adds non-linearity

Removes negative values

Speeds up training

# 3️⃣ Pooling Layer

Reduces feature map size

Keeps important information

Types:

Max Pooling (most common)

Average Pooling

# 4️⃣ Fully Connected Layer

Flattens features

Performs classification

# 🔹 Key Concepts (Very Important)
🔸 Local Receptive Field

Each neuron looks at a small part of the image.

🔸 Weight Sharing

Same filter is used across the image → fewer parameters.

🔸 Spatial Hierarchy

Early layers → edges

Middle layers → shapes

Deep layers → objects

# 🔹 CNN vs ANN
| Feature      | ANN          | CNN         |
| ------------ | ------------ | ----------- |
| Input        | Flattened    | 2D/3D image |
| Parameters   | Very large   | Fewer       |
| Spatial info | Lost         | Preserved   |
| Best for     | Tabular data | Images      |

# 🔹 Applications of CNN

📸 Image Classification
🧠 Face Recognition
🚗 Self-driving cars
🩺 Medical Imaging
📦 Object Detection

🧠 One-Line Exam Answer

A Convolutional Neural Network is a deep learning model that uses convolution operations to automatically learn spatial features from images.
