# 🔹 What Is Hyperparameter Tuning?

Hyperparameters are values you set before training, such as:

Learning rate

Number of layers

Number of neurons

Batch size

Optimizer

Dropout rate

👉 Hyperparameter tuning means finding the best combination of these values to get maximum performance.

# 🔹 What Is Keras Tuner?

Keras Tuner is a library that automatically searches for the best hyperparameters for your neural network.

✔ Saves time
✔ Better accuracy
✔ No manual trial & error

# 🔹 Why Not Manual Tuning?

Manual tuning is:

Slow

Guess-based

Not scalable

Keras Tuner does it systematically ✅

# 🔹 What Can Keras Tuner Tune?

Number of hidden layers

Neurons per layer

Activation functions

Optimizers

Learning rate

Dropout rate

# 🔹 Main Search Strategies
1️⃣ Random Search (Most Used)

Tries random combinations

Faster than Grid Search

2️⃣ Hyperband ⭐

Smart & fast

Stops bad models early

3️⃣ Bayesian Optimization

Uses past results to choose next values

More intelligent, slower

# 🔹 Basic Workflow (Very Important)
Define model →
Define hyperparameters →
Choose tuner →
Search →
Get best model

# 🔹 Example: Keras Tuner (Simple ANN)
1️⃣ Install Keras Tuner
pip install keras-tuner

2️⃣ Import Libraries
import keras_tuner as kt
from tensorflow import keras
from tensorflow.keras import layers

3️⃣ Build Model Function
def build_model(hp):
    model = keras.Sequential()

    # Tune number of units
    units = hp.Int('units', min_value=32, max_value=256, step=32)
    model.add(layers.Dense(units, activation='relu'))

    # Tune dropout
    dropout = hp.Float('dropout', 0.1, 0.5, step=0.1)
    model.add(layers.Dropout(dropout))

    model.add(layers.Dense(1, activation='sigmoid'))

    # Tune learning rate
    lr = hp.Choice('learning_rate', [1e-2, 1e-3, 1e-4])
    model.compile(
        optimizer=keras.optimizers.Adam(lr),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )
    return model

4️⃣ Create Tuner
tuner = kt.RandomSearch(
    build_model,
    objective='val_accuracy',
    max_trials=5,
    directory='my_dir',
    project_name='ann_tuning'
)

5️⃣ Start Search
tuner.search(
    X_train, y_train,
    epochs=10,
    validation_data=(X_val, y_val)
)

6️⃣ Get Best Model
best_model = tuner.get_best_models(num_models=1)[0]

🔹 Key Parameters Explained

| Parameter     | Meaning            |
| ------------- | ------------------ |
| `objective`   | Metric to optimize |
| `max_trials`  | Number of models   |
| `epochs`      | Training epochs    |
| `hp.Int()`    | Integer range      |
| `hp.Float()`  | Float range        |
| `hp.Choice()` | Fixed options      |
