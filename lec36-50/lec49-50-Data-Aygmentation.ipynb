Data Augmentation (Deep Learning / CNN)
# What is Data Augmentation?

Data Augmentation means:

Artificially increasing training data by modifying existing images

⚠️ We don’t collect new images — we create variations.

# Why Data Augmentation is Needed

CNNs need lots of data.
When data is limited, models:
❌ Overfit
❌ Memorize images

Augmentation helps:
✅ Improve generalization
✅ Reduce overfitting
✅ Make model robust

# Example (Easy)

One cat image →

Rotated

Flipped

Zoomed

Shifted

Brightness changed

👉 CNN thinks they are new images

# Common Data Augmentation Techniques
| Technique          | Meaning                |
| ------------------ | ---------------------- |
| Rotation           | Rotate image           |
| Horizontal flip    | Mirror image           |
| Width/Height shift | Move image             |
| Zoom               | Zoom in/out            |
| Shear              | Tilt image             |
| Brightness         | Light/dark             |
| Rescale            | Normalize pixel values |

Important Rule ⭐

❗ Apply augmentation ONLY on training data
❌ Never on validation or test data


# How Augmentation Works Internally

Images are augmented on the fly

No new images are stored on disk

Every epoch sees new variations

# Data Augmentation vs Regularization
| Feature     | Data Augmentation  | Dropout            |
| ----------- | ------------------ | ------------------ |
| Acts on     | Data               | Network            |
| Purpose     | Reduce overfitting | Reduce overfitting |
| Storage     | No extra storage   | No                 |
| Used in CNN | ⭐⭐⭐                | ⭐⭐                 |

🧠 One-Line Exam Answer

Data augmentation artificially increases training data by applying transformations to improve model generalization.

💡 Interview Questions

❓ Should we apply augmentation to test data?
❌ NO

❓ Does augmentation create real data?
❌ NO (synthetic)
