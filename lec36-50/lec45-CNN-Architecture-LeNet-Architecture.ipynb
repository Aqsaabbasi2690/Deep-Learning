# 🔹 What Is CNN Architecture?

A CNN architecture defines how convolutional, pooling, and fully connected layers are arranged to process images.

Layers are stacked hierarchically

Early layers → simple features (edges)

Deeper layers → complex features (objects)

# 🔹 LeNet-5 Overview

LeNet-5 is a classic CNN designed by Yann LeCun in 1998 for handwritten digit recognition (MNIST dataset).

Input: 32×32 grayscale image

Output: 10 classes (digits 0-9)

# 🔹 LeNet-5 Architecture Step-by-Step

| Layer | Type                      | Output Size | Notes                                             |
| ----- | ------------------------- | ----------- | ------------------------------------------------- |
| 1     | Input                     | 32×32       | Grayscale image                                   |
| 2     | C1: Convolution           | 28×28×6     | 6 filters, 5×5 kernel, stride 1, activation: tanh |
| 3     | S2: Subsampling (Pooling) | 14×14×6     | Average pooling, 2×2, stride 2                    |
| 4     | C3: Convolution           | 10×10×16    | 16 filters, 5×5 kernel                            |
| 5     | S4: Subsampling (Pooling) | 5×5×16      | Average pooling, 2×2, stride 2                    |
| 6     | C5: Convolution           | 1×1×120     | Fully connected convolution                       |
| 7     | F6: Fully Connected       | 84          | Activation: tanh                                  |
| 8     | Output Layer              | 10          | Softmax (classification)                          |


# 🔹 Key Features of LeNet-5

Small filters (5×5) → captures local features

Subsampling (Pooling) → reduces size

Tanh activation → original CNN (now replaced by ReLU)

Fully connected layer → final classification

# 🔹 Visualization (Simplified)
Input 32x32
     ↓ Conv 5x5 → 6 feature maps 28x28
     ↓ Avg Pool 2x2 → 6 feature maps 14x14
     ↓ Conv 5x5 → 16 feature maps 10x10
     ↓ Avg Pool 2x2 → 16 feature maps 5x5
     ↓ Conv 5x5 → 120
     ↓ FC → 84
     ↓ Output → 10 classes

# 🔹 LeNet-5 vs Modern CNNs
| Feature    | LeNet-5     | Modern CNN (e.g., VGG, ResNet) |
| ---------- | ----------- | ------------------------------ |
| Layers     | 7           | 20+                            |
| Activation | Tanh        | ReLU                           |
| Pooling    | Average     | Max / Global Avg               |
| Input Size | 32×32       | 224×224+                       |
| Use        | Handwriting | Complex images                 |

# 🔹 Why LeNet-5 Is Important

First successful CNN for images

Introduced convolution + pooling layers

Inspired modern architectures

🧠 One-Line Exam Answer

LeNet-5 is a classic CNN architecture with convolutional and pooling layers, designed for handwritten digit recognition.

💡 Interview Tip

❓ Why pooling in LeNet-5?
✔ Reduces feature map size
✔ Provides translation invariance
✔ Prevents overfitting
