# 🔹 What Is AdaGrad?

AdaGrad is an optimizer that adapts the learning rate for each parameter based on past gradients.

👉 Frequently updated parameters get smaller learning rates,
👉 Rarely updated parameters get larger learning rates.

# 🔹 Why AdaGrad Was Introduced

Problem with SGD:

Same learning rate for all parameters

Not good for sparse data

AdaGrad fixes this by:

Automatically adjusting learning rates

# 🔹 Core Idea (Intuition)

Think of learning like walking downhill:

Steep path → slow down

Flat path → move faster

AdaGrad remembers past gradients to adjust step size.

# 🔹 Mathematical Formulation
Accumulate Squared Gradients
Gₜ = Gₜ₋₁ + gₜ²

Weight Update
wₜ = wₜ₋₁ − η · gₜ / (√Gₜ + ε)


Where:

gₜ = gradient at time t

η = learning rate

ε = small constant for stability

# 🔹 Key Characteristics

| Feature       | AdaGrad             |
| ------------- | ------------------- |
| Learning rate | Adaptive            |
| Memory        | All past gradients  |
| Best for      | Sparse data         |
| Convergence   | Fast initially      |
| Long training | ❌ Learning rate → 0 |



# 🔹 Advantages

✅ No manual learning rate tuning
✅ Works well with sparse features
✅ Simple to implement

# 🔹 Disadvantages (Very Important ❗)

❌ Learning rate keeps decreasing
❌ Training may stop early
❌ Not suitable for deep networks


👉 This is why RMSProp and Adam were introduced.

| Feature       | SGD   | AdaGrad  |
| ------------- | ----- | -------- |
| Learning rate | Fixed | Adaptive |
| Sparse data   | ❌     | ✅        |
| Long training | ✅     | ❌        |


# 🔹 AdaGrad vs RMSProp

| AdaGrad                   | RMSProp              |
| ------------------------- | -------------------- |
| Accumulates all gradients | Uses moving average  |
| Learning rate → 0         | Stable learning rate |
| Older                     | Improved             |

🧠 One-Line Exam Answer

AdaGrad adapts the learning rate of each parameter by scaling updates using the accumulated squared gradients.
