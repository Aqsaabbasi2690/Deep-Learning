How Backpropagation Works in CNN Layers

We’ll cover only these three, exactly as you asked:

Convolution layer

MaxPooling layer

Flatten layer

🔁 Big Picture First

Backpropagation means:

Error flows from output → back to input → weights get updated

Forward:  Image → Conv → MaxPool → Flatten → FC → Output
Backward: Output → FC → Flatten → MaxPool → Conv

1️⃣ Backpropagation in Convolution Layer
🔹 Non-Technical Intuition

Convolution layer has filters (kernels)

Each filter looks for a pattern (edge, curve, shape)

If prediction is wrong:
👉 Filters that caused the mistake are adjusted

Think:

“This filter reacted too much — reduce it next time”

🔹 What Gets Updated?

✅ Filter weights
✅ Bias
❌ Input image (not learnable)

🔹 Technical Explanation (Simple)

During backprop:

Compute gradient of loss w.r.t filter weights

Use chain rule

Slide error over the input (just like convolution)

Weight update:
W_new = W_old − learning_rate × dLoss/dW


📌 Important:
Because of weight sharing, one filter update affects the whole image.

🧠 Key Point (Exam Favorite ⭐)

Backpropagation in convolution updates filters using shared gradients.

2️⃣ Backpropagation in MaxPooling Layer
🔹 Non-Technical Intuition

MaxPooling chooses the biggest value

During backprop:
👉 Error goes only to that biggest value

Think:

“Only the winner gets the blame”

🔹 Does MaxPooling Have Weights?

❌ NO
❌ Nothing to update

It only routes gradients

🔹 Technical Explanation

Example:

2   5
1   7   ← max


Backward pass:

Gradient goes only to 7

Others get zero gradient

📌 CNN remembers max index (argmax) during forward pass.

🧠 Key Point

MaxPooling backprop passes gradients only to maximum values.

3️⃣ Backpropagation in Flatten Layer
🔹 Non-Technical Intuition

Flatten layer:

Just reshapes data

No learning, no decision

Think:

“I only changed the shape, not the meaning”

🔹 What Happens in Backprop?

Gradient is reshaped back

Values stay the same

Shape changes only

🔹 Technical Explanation

Forward:

(5 × 5 × 16) → (400)


Backward:

(400) → (5 × 5 × 16)


📌 No parameters, no gradients to compute.

🧠 Key Point

Flatten layer only reshapes gradients during backpropagation.

🔥 Summary Table (Very Important)


| Layer       | Learnable Weights? | Backprop Action    |
| ----------- | ------------------ | ------------------ |
| Convolution | ✅ Yes              | Updates filters    |
| MaxPooling  | ❌ No               | Routes gradients   |
| Flatten     | ❌ No               | Reshapes gradients |


🧩 Memory Trick (Lock This In 🔒)
Conv → Learn
MaxPool → Select
Flatten → Reshape

🎯 One-Line Exam Answers

Convolution: Backprop updates filters using shared gradients

MaxPooling: Backprop passes error only to max values

Flatten: Backprop reshapes gradients without learning
