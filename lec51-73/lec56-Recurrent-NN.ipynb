Recurrent Neural Network (RNN)

An RNN is a neural network designed for sequential data — data where order matters.

Examples:

Text / sentences

Time series (stock prices, weather)

Speech / audio

Video frames

👉 ANN & CNN don’t remember previous inputs.
👉 RNN remembers past information.

🧠 Why is it called “Recurrent”?

Because the network feeds its output back into itself at the next time step.

So instead of:

Input → Output


We have:

Input → Hidden → Output
           ↑
        (memory)


That loop = recurrence

🏗️ RNN Architecture (Basic)

At time step t:

xₜ → [ RNN Cell ] → hₜ → yₜ
          ↑
        hₜ₋₁


Where:

xₜ = input at time t

hₜ₋₁ = previous hidden state (memory)

hₜ = current hidden state

yₜ = output

💡 Hidden state = memory of the past

🔁 Forward Propagation in RNN (Step-by-Step)

Let’s say input is a sentence:

“I love AI”

We process one word at a time.

Step 1: First word
x₁ = "I"
h₁ = f(Wₓx₁ + Wₕh₀ + b)


(h₀ is usually zero)

Step 2: Second word
x₂ = "love"
h₂ = f(Wₓx₂ + Wₕh₁ + b)

Step 3: Third word
x₃ = "AI"
h₃ = f(Wₓx₃ + Wₕh₂ + b)


➡️ Each step uses previous memory

📐 Mathematical Formula (Simple)

Hidden state:

hₜ = tanh(Wₓxₜ + Wₕhₜ₋₁ + b)


Output:

yₜ = softmax(Wᵧhₜ)


Activation:

tanh → keeps values between -1 and 1

softmax → for classification

⏳ Types of RNN Architectures
1️⃣ One-to-One

Image → Label (ANN style)

2️⃣ One-to-Many

Image → Caption

3️⃣ Many-to-One

Sentence → Sentiment (Positive / Negative)

4️⃣ Many-to-Many

Translation

Speech recognition

⚠️ Problems with Basic RNN

❌ Vanishing Gradient
❌ Can’t remember long sequences

👉 That’s why we use:

LSTM (Long Short-Term Memory)

GRU (Gated Recurrent Unit)

🧩 Summary (Exam + Interview Ready)

RNN is used for sequential data

It has memory via hidden states

Forward propagation happens time-step by time-step

Same weights are reused at every step

Struggles with long-term dependencies → solved by LSTM & GRU
