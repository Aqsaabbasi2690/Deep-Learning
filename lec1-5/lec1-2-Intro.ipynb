# What is Deep Learning?

Deep Learning uses artificial neural networks inspired by the human brain.
These networks have many layers (that’s why it’s called deep).


Mostly ML algos focus on statistical techniques, dl focus on logical structure. which is called neural networks(ispired by human brain) these collections of algos are called dl.


👉 Instead of manually programming rules, deep learning learns patterns automatically from data.

Simple idea:

Input data (images, text, audio)

Goes through many layers of neurons

Each layer learns more complex features

Final output = prediction or decision

Example:

First layer → detects edges in an image

Middle layers → detect shapes (eyes, nose)

Final layer → recognizes a face

# 🔹 2. Convolutional Neural Network (CNN)

Best for images & videos

📌 How it works:

Uses filters (kernels) to detect patterns

Learns edges → shapes → objects

✅ Used for:

Image recognition

Face detection

Medical imaging

Video analysis

📍 Examples:

Face unlock 📱

Object detection in self-driving cars 🚗

# 🔹 3. Recurrent Neural Network (RNN)

Designed for sequential data

📌 How it works:

Has memory

Previous output affects current input

✅ Used for:

Text

Speech

Time-series data

📍 Examples:

Sentence prediction

Stock price prediction

⚠️ Problem: Vanishing gradient (can forget long-term info)

# 🔹 4. LSTM (Long Short-Term Memory)

Improved version of RNN

📌 How it works:

Remembers long-term dependencies

Uses gates (forget, input, output)

✅ Used for:

Language translation

Speech recognition

Chatbots (older ones)

# 🔹 5. GRU (Gated Recurrent Unit)

Simpler than LSTM

📌 How it works:

Fewer gates

Faster training

✅ Used for:

NLP tasks

Time-series forecasting

# 🔹 6. Autoencoders

Unsupervised learning model

📌 How it works:

Compresses data → reconstructs it

Learns important features

✅ Used for:

Dimensionality reduction

Noise removal

Anomaly detection

# 🔹 7. Generative Adversarial Networks (GANs)

Two networks competing

📌 How it works:

Generator creates fake data

Discriminator detects fake vs real

✅ Used for:

Image generation

Face synthesis

Art & deepfakes

# 🔹 8. Transformer Networks 🔥

Most powerful & modern

📌 How it works:

Uses attention mechanism

Processes data in parallel

No recurrence needed

✅ Used for:

NLP

GenAI

Large language models

📍 Examples:

ChatGPT

Google Translate

BERT, GPT, T5

# 🔹 9. Graph Neural Networks (GNN)

Works with graph data

📌 How it works:

Nodes + edges

Learns relationships

✅ Used for:

Social networks

Recommendation systems

Fraud detection

# 🔹 10. Spiking Neural Networks (SNN)

Brain-inspired (advanced research)

📌 How it works:

Neurons fire at specific times

Energy-efficient

✅ Used for:

Neuromorphic computing

Robotics

Brain simulation
