It is an algorithm it is use for supervised machine learning. It is also called a mathematical model/function.

Perceptron is inspired with neuron 

Percepton only classify linear or sort of linear data. Perceptron is a line in 2d , which creates two regions / 2 classes/ 2 classify. in 3d its become plane and 4 d its will become hyperplane. 

A perceptron takes some information, thinks about it using simple rules, and gives one decision.

How a Perceptron Thinks (No Tech Words)

1️⃣ Receives information
Example: marks, attendance, behavior

2️⃣ Gives importance to each thing
Some inputs matter more than others

3️⃣ Adds everything together

4️⃣ Makes a decision
YES or NO
PASS or FAIL
SPAM or NOT SPAM

⚠️ Limitation (Important)

A single perceptron:
❌ Cannot understand complex patterns
❌ Cannot solve problems like:

Face recognition

Language understanding

Images with curves or overlaps

👉 That’s why multi-layer perceptrons were created.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv('data/placement.csv')
print(df.shape)
df.head()
```

```python
sns.scatterplot(
    x='cgpa',
    y='resume_score',
    hue='placed',
    data=df
)
```

```python
X = df.iloc[:,0:2]
y = df.iloc[:,-1]
 
from sklearn.linear_model import Perceptron
p = Perceptron()
p.fit(X,y)
```


Perceptron()
