#🔹 Why ReLU Variants Exist

ReLU problem:

Neurons can “die” when inputs become negative → gradient = 0

👉 ReLU variants fix dying ReLU and improve learning stability


# 1️⃣ Leaky ReLU

f(x) = x        if x > 0
f(x) = αx       if x ≤ 0   (α ≈ 0.01)


🧠 Idea

Allows small negative output

Gradient never becomes zero

✅ Advantages

✔ Fixes dying ReLU
✔ Simple & fast

❌ Disadvantage

✘ α is fixed (not learned)

📌 Used in hidden layers

# 2️⃣ Parametric ReLU (PReLU)

f(x) = x        if x > 0
f(x) = αx       if x ≤ 0   (α is learned)

🧠 Idea

Network learns the best α

✅ Advantages

✔ More flexible than Leaky ReLU
✔ Better accuracy in deep networks

❌ Disadvantage

✘ More parameters → risk of overfitting


# 3️⃣ ELU (Exponential Linear Unit)

f(x) = x                    if x > 0
f(x) = α (e^x − 1)           if x ≤ 0

🧠 Idea

Smooth curve for negative values

Pushes mean activation closer to zero

✅ Advantages

✔ Faster convergence
✔ Reduces bias shift
✔ Smooth gradient

❌ Disadvantages

✘ Computationally expensive
✘ Slower than ReLU


# 4️⃣ SELU (Scaled ELU)

SELU(x) = λ * ELU(x)


(λ and α are fixed constants)

🧠 Idea

Self-Normalizing Neural Networks

Keeps activations:

Mean ≈ 0

Variance ≈ 1

✅ Advantages

✔ No need for Batch Normalization
✔ Stable deep networks

❌ Important Conditions ⚠️

❗ Use LeCun Normal initialization
❗ Use AlphaDropout (not normal dropout)

# When to Use What

Default choice → ReLU

Dying ReLU problem → Leaky ReLU

Very deep network → ELU / SELU

No BatchNorm wanted → SELU

🧠 One-Line Exam Answers

Leaky ReLU: Allows small negative gradient

PReLU: Learns negative slope

ELU: Smooth exponential negative part

SELU: Self-normalizes activations
