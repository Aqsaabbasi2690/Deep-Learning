# Why Weight Initialization Matters

Weights decide:

How fast a network learns

Whether gradients vanish or explode

Whether neurons learn different features

❌ Bad initialization = no learning

# 🚫 What NOT to Do (Critical)

# ❌ 1. Initialize All Weights to Zero

Problem:

All neurons learn the same thing

No symmetry breaking

👉 Network becomes useless

# ❌ 2. Use Very Large Random Values

w ~ N(0, 100)

Problem:

Exploding gradients

Activation saturation (sigmoid/tanh)


# ❌ 3. Use Very Small Values
w ~ N(0, 0.0001)


Problem:

Vanishing gradients

Slow learning

# ❌ 4. Same Initialization for Every Layer

Different layers need different scaling

Causes unstable training

# ❌ 5. Ignoring Activation Function

Initialization must match activation

ReLU ≠ Sigmoid

# ✅ Good Weight Initialization Techniques
# 1️⃣ Random Initialization (Basic)
w ~ Uniform(−0.01, 0.01)


✔ Breaks symmetry
❌ Not suitable for deep networks

# 2️⃣ Xavier (Glorot) Initialization
📌 Best for:

Sigmoid

Tanh

📐 Formula
Var(w) = 1 / n


(or 2 / (n_in + n_out))

✅ Benefit

Prevents vanishing gradients

Keeps activations stable

# 3️⃣ He Initialization ⭐ (Most Important)
📌 Best for:

ReLU

Leaky ReLU

📐 Formula
Var(w) = 2 / n

✅ Benefit

Prevents dying ReLU

Faster convergence

# 4️⃣ LeCun Initialization
📌 Best for:

SELU

📐 Formula
Var(w) = 1 / n


| Initialization | Activation        | Used For            |
| -------------- | ----------------- | ------------------- |
| Random         | Any               | Shallow networks    |
| Xavier         | Sigmoid / Tanh    | ANN                 |
| He             | ReLU / Leaky ReLU | Deep CNN/ANN        |
| LeCun          | SELU              | Self-normalizing NN |


Golden Rule (Must Remember)

Match weight initialization with activation function



Weight initialization assigns initial values to neural network parameters in a way that ensures stable and efficient training.

💡 Interview Question

❓ Why zero initialization fails?
✔ Breaks symmetry
✔ All neurons update identically


#  Simple Mapping (Memorize This)
ReLU  → He
Sigmoid/Tanh → Xavier
SELU → LeCun
