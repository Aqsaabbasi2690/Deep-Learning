# 🔹 What Is Nesterov Accelerated Gradient?

NAG is an improved version of SGD with Momentum that looks ahead before updating weights.

👉 It corrects the update before overshooting the minimum.

#🔹 Why NAG Was Introduced

SGD with Momentum problem:

Momentum may overshoot the minimum

Update uses gradient at current position

NAG solution:

Compute gradient at the future position

Make smarter corrections

# 🔹 Core Idea (Simple Intuition)

Imagine walking downhill 👣:

Momentum: take a big step, then adjust

NAG: look ahead, then decide how big the step should be

# 🔹 Mathematical Formulation
Step 1: Look-Ahead Position
w_lookahead = w − βv

Step 2: Compute Gradient at Look-Ahead
g = ∇L(w_lookahead)

Step 3: Update Velocity
v = βv + ηg

Step 4: Update Weights
w = w − v


Where:

η = learning rate

β = momentum (usually 0.9)

#🔹 NAG vs Momentum (Very Important)

| Feature              | Momentum         | NAG                 |
| -------------------- | ---------------- | ------------------- |
| Gradient computed at | Current position | Look-ahead position |
| Overshooting         | Possible         | Reduced             |
| Convergence          | Fast             | ⭐ Faster            |
| Stability            | Medium           | ⭐ High              |

# 🔹 Advantages of NAG

✅ Faster convergence
✅ Better control near minima
✅ Less oscillation
✅ More accurate updates


# 🔹 Disadvantages

❌ Slightly more complex
❌ Extra computation (minor)

#🔹 NAG in Keras
from tensorflow.keras.optimizers import SGD

optimizer = SGD(
    learning_rate=0.01,
    momentum=0.9,
    nesterov=True
)

model.compile(
    optimizer=optimizer,
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

🧠 One-Line Exam Answer

Nesterov Accelerated Gradient improves momentum by computing gradients at a look-ahead position, resulting in faster and more stable convergence.
