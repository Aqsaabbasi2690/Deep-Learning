# What Is Batch Normalization (BN)?

Batch Normalization is a technique that normalizes the activations of a layer for each 

mini-batch during training.

It makes training faster, more stable, and more accurate.

# 🔹 Why Do We Need Batch Normalization?

During training, weights keep changing →

the distribution of activations also keeps changing (internal covariate shift).

👉 BN fixes this by keeping activations well-scaled.

# 🔹 How Batch Normalization Works (Step-by-Step)

For each mini-batch:

1️⃣ Compute mean

2️⃣ Compute variance

3️⃣ Normalize

4️⃣ Scale & shift (learnable parameters γ and β)

Mathematically:

x̂ = (x − μ) / √(σ² + ε)

y = γx̂ + β

# 🔹 Where Is Batch Normalization Applied?

Usually:

Dense/Conv → BatchNorm → Activation


✔ Before activation

✔ After linear transformation

# 🔹 Benefits of Batch Normalization

✅ Faster convergence

✅ Higher learning rates allowed

✅ Reduces vanishing/exploding gradients

✅ Acts as regularization

✅ Less sensitive to weight initialization

# 🔹 Batch Normalization vs Feature Scaling

Feature	Feature Scaling	Batch Normalization

Applied to	Input data	Hidden layers

Happens	Before training	During training

Learnable params	❌	✅ (γ, β)
# 🔹 Batch Normalization vs Dropout

| Batch Norm             | Dropout                  |
| ---------------------- | ------------------------ |
| Normalizes activations | Randomly removes neurons |
| Speeds up training     | Slows training           |
| Less noise             | Adds noise               |
| Often replaces dropout | Used when overfitting    |


# 🔹 Batch Learning in Keras
📌 What Is Batch Learning?

Batch learning means training the model using batches of data instead of the whole dataset at once.

Types:

Batch Gradient Descent

Mini-Batch Gradient Descent ⭐ (default)

Stochastic Gradient Descent

# 🔹 Batch Size in Keras
model.fit(X_train, y_train, batch_size=32, epochs=20)

Common batch sizes:

16, 32, 64, 128

# 🔹 Adding Batch Normalization in Keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, BatchNormalization, Activation

model = Sequential([
    Dense(64, input_shape=(100,)),
    BatchNormalization(),
    Activation('relu'),
    Dense(1, activation='sigmoid')
])

# 🔹 Important Training vs Inference Difference

| Phase    | Batch Norm Behavior          |
| -------- | ---------------------------- |
| Training | Uses batch mean & variance   |
| Testing  | Uses running mean & variance |
