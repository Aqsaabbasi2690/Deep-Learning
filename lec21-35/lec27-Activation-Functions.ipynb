If we not apply activation function in our neural network, then neural network not capture non linear data, only will capture linear data.

#  Activation Function 

An activation function decides whether a neuron should “fire” or not by transforming the input signal into an output.

#🔹 Why Activation Function Is Needed

Without activation functions:

Neural networks become linear

They cannot learn complex patterns

👉 Activation functions add non-linearity, making deep learning powerful.


# What Happens in a Neuron
Input → Weighted Sum → Activation Function → Output


Mathematically:

z = w·x + b
a = f(z)



#  Common Activation Functions
1️⃣ Step Function (Perceptron)

📌 Output: 0 or 1

❌ Not used in deep learning
❌ No gradient → cannot train with backpropagation

2️⃣ Sigmoid

📐 Range: (0, 1)

✅ Advantages

Used in binary classification

Probabilistic output

❌ Disadvantages

Vanishing gradient

Slow training

📌 Used in output layer only

3️⃣ Tanh

📐 Range: (−1, 1)

✅ Better than sigmoid

Zero-centered

❌ Still suffers from vanishing gradient
4️⃣ ReLU (Most Popular ⭐)

📐 f(x) = max(0, x)

✅ Advantages

Fast training

No vanishing gradient (for +ve values)

❌ Problem

Dying ReLU (neurons stuck at 0)

📌 Used in hidden layers

5️⃣ Leaky ReLU

📐 Allows small negative values

✅ Fixes dying ReLU problem
6️⃣ Softmax

📐 Range: (0, 1), sum = 1

📌 Used for:

Multi-class classification

Output layer


| Function   | Range         | Used In            |
| ---------- | ------------- | ------------------ |
| Sigmoid    | 0–1           | Binary output      |
| Tanh       | −1 to 1       | Hidden layers      |
| ReLU       | 0–∞           | Hidden layers      |
| Leaky ReLU | −∞–∞          | Hidden layers      |
| Softmax    | Probabilities | Multi-class output |


# 🎯 Which Activation Function to Use?

| Layer              | Activation        |
| ------------------ | ----------------- |
| Hidden layers      | ReLU / Leaky ReLU |
| Binary output      | Sigmoid           |
| Multi-class output | Softmax           |
| Regression output  | Linear            |


An activation function introduces non-linearity into a neural network and decides whether a neuron should be activated.
