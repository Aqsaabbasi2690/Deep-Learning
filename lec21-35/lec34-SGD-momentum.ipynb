# 🔹 What Is SGD with Momentum?

SGD with Momentum is an improved version of Stochastic Gradient Descent that accelerates learning by accumulating past gradients.

👉 It helps the optimizer move faster in the right direction and reduces oscillations.

# 🔹 Why Do We Need Momentum?

Plain SGD problems:

Noisy updates

Zig-zag movement

Slow convergence

Momentum fixes this by:

Remembering past gradients

Smoothing updates

# 🔹 Intuition (Easy Example)

Imagine a ball rolling downhill 🏀

Gravity = gradient

Speed = momentum

Once moving in one direction, it keeps moving faster.

# 🔹 Mathematical Formulation

Velocity Update (EWMA of gradients)


vₜ = β vₜ₋₁ + (1 − β) gₜ

Weight Update


wₜ = wₜ₋₁ − η vₜ

Where:

gₜ = current gradient

η = learning rate

β = momentum coefficient (usually 0.9)

# 🔹 Role of β (Momentum Term)

β Value	Effect

0	Same as SGD

0.9	Fast & smooth (best choice)

0.99	Very smooth but slower reaction

# 🔹 Advantages
✅ Faster convergence
✅ Reduced oscillations
✅ Escapes shallow local minima
✅ Works well for deep networks

🔹 Disadvantages
❌ Requires tuning β
❌ May overshoot if learning rate is large

# 🔹 SGD vs SGD with Momentum

| Feature     | SGD    | SGD + Momentum |
| ----------- | ------ | -------------- |
| Speed       | Slow   | ⭐ Fast         |
| Oscillation | High   | Low            |
| Stability   | Medium | High           |
| Memory      | No     | Yes (velocity) |


# 🔹 SGD with Momentum in Keras
from tensorflow.keras.optimizers import SGD

optimizer = SGD(learning_rate=0.01, momentum=0.9)

model.compile(
    optimizer=optimizer,
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

✔ Deep networks
✔ When Adam overfits
✔ Large datasets

🧠 One-Line Exam Answer

SGD with Momentum accelerates gradient descent by accumulating an exponentially weighted moving average of past gradients.
