# 🔹 What Is an Optimizer?

An optimizer updates the weights of a neural network to minimize the loss function.

👉 It controls how and how fast a model learns.

# 🔹 Why Optimizers Matter

Faster convergence

Stable training

Avoids local minima & saddle points

Bad optimizer ❌ → slow or failed training

# 🔥 Main Types of Optimizers
# 1️⃣ Gradient Descent (GD)
📌 Idea

Update weights using full dataset.

📐 Update Rule
w = w − η ∇L

❌ Problems

Very slow

High memory usage

# 2️⃣ Stochastic Gradient Descent (SGD)
📌 Idea

Update weights using one sample at a time.

✅ Pros

Faster updates

Escapes local minima

❌ Cons

Noisy updates

Unstable convergence

# 3️⃣ Mini-Batch Gradient Descent ⭐
📌 Idea

Update weights using small batches (32, 64).

✅ Pros

Best balance

Used in practice

# 4️⃣ SGD with Momentum
🧠 Idea

Uses past gradients to accelerate learning.

📐 Update Rule
v = βv + η∇L
w = w − v

✅ Benefits

✔ Faster
✔ Less oscillation

# 5️⃣ AdaGrad
🧠 Idea

Adaptive learning rate for each parameter.

❌ Problem

Learning rate becomes too small.

3 6️⃣ RMSProp
🧠 Idea

Fixes AdaGrad’s problem using moving average.

✅ Good for:

RNNs

Non-stationary problems

# 7️⃣ Adam ⭐ (Most Popular)
🧠 Idea

Combines:

Momentum

RMSProp

📐 Uses:

First moment (mean)

Second moment (variance)

✅ Advantages

✔ Fast convergence
✔ Works well out-of-the-box
✔ Handles sparse gradients

# 8️⃣ AdamW
📌 Improvement over Adam

Decouples weight decay

Better generalization

# 🔄 Optimizer Comparison Table

| Optimizer      | Learning Rate | Speed  | Stability |
| -------------- | ------------- | ------ | --------- |
| SGD            | Fixed         | Slow   | Medium    |
| SGD + Momentum | Fixed         | Faster | High      |
| RMSProp        | Adaptive      | Fast   | High      |
| Adam           | Adaptive      | ⭐ Fast | ⭐ High    |
| AdamW          | Adaptive      | ⭐ Fast | ⭐ High    |



# 🎯 Which Optimizer to Use?

| Scenario     | Optimizer      |
| ------------ | -------------- |
| Simple ANN   | SGD + Momentum |
| CNN          | Adam / AdamW   |
| RNN          | RMSProp / Adam |
| Transformers | AdamW          |
| Sparse data  | Adam           |


One-Line Exam Answer

An optimizer is an algorithm that updates neural network weights to minimize the loss function efficiently.
