# 🔹 What Is EWMA?

EWMA is a technique that computes an average where recent values are given more 
importance than older ones.

👉 Used to smooth noisy data and track trends.

# 🔹 Why Is It Used in Deep Learning?

Gradients are noisy, especially in SGD.

EWMA helps to:

Smooth gradient updates

Reduce oscillations

Speed up convergence

📌 That’s why it’s used in Momentum, RMSProp, Adam

# 🔹 EWMA Formula

vₜ = β · vₜ₋₁ + (1 − β) · xₜ


Where:

xₜ = current value (e.g., gradient)

vₜ = EWMA value

β = smoothing parameter (0 < β < 1)

# 🔹 Role of β (Very Important)

β Value	Effect

0.9	Smooth (uses last ~10 values)

0.99	Very smooth

Small β	More noisy

Large β	More memory

📌 Effective window size ≈ 1 / (1 − β)

# 🔹 EWMA in Optimizers

# 1️⃣ Momentum

v = βv + (1−β)g

w = w − v


👉 Smooths gradients

# 2️⃣ RMSProp

s = βs + (1−β)g²

w = w − η g / √(s + ε)


👉 Controls learning rate

# 3️⃣ Adam (Best Example)

Uses two EWMAs:

Mean of gradients

Variance of gradients

m = β₁m + (1−β₁)g

v = β₂v + (1−β₂)g²

# 🔹 Bias Correction (Important)

Early steps are biased toward zero.

Fix:
m̂ = m / (1 − βᵗ)


Used in Adam optimizer

# 🔹 EWMA vs Simple Moving Average

| Feature       | EWMA                | Simple MA    |
| ------------- | ------------------- | ------------ |
| Weights       | Exponential         | Equal        |
| Recent values | More important      | Same         |
| Memory        | Infinite (decaying) | Fixed window |
| Smoothness    | High                | Medium       |

🧠 One-Line Exam Answer

Exponentially weighted moving average gives more importance to recent values and is used in optimizers to smooth gradients.
